```python
# SETUP

import pandas as pd
import numpy as np

df=pd.read_csv('/content/Final Dataset - State of Data 2024 - Kaggle - df_survey_2024.csv')
```

```python
for x in df:
    print(x)
```

```python
# SELEÇÃO COLUNAS

cols_p1 = {
  '1.a_idade':'idade',
  '1.b_genero': 'genero',
  '1.c_cor/raca/etnia': 'cor',
  '1.l_nivel_de_ensino': 'escolaridade',
  '1.m_área_de_formação': 'formacao',
  '1.i_estado_onde_mora': 'uf_mora',
  '1.j_vive_no_estado_de_formacao': 'permaneceu_uf',
  '1.k_estado_de_origem': 'uf_origem',
  '2.a_situação_de_trabalho': 'situacao',
  '2.f_cargo_atual': 'cargo',
  '2.g_nivel': 'nivel',
  '2.h_faixa_salarial': 'salario',
  '2.i_tempo_de_experiencia_em_dados': 'tempo_dados',
  '2.j_tempo_de_experiencia_em_ti': 'tempo_ti'
}

# RENOMEAR COLUNAS

df_p1 = df[cols_p1.keys()]
df_p1 = df_p1.rename(columns=cols_p1)

df_p1.info()
df_p1.head()
```

```python
# TRATAMENTO

# ajuste 1
# preencher uf origem a partir da outra coluna

# 1.1 verifica se bool ou string
# 1.2 se permaneceu_uf == true, uf_origem = uf_mora

# 1.1
df_p1['permaneceu_uf'].apply(type).value_counts()
```

```python
# 1.2
df_p1.loc[df_p1['permaneceu_uf'] == True, 'uf_origem'] = df_p1['uf_mora']

df_p1['uf_mora'].info()
df_p1['uf_origem'].info()
```

```python
# ajuste 2
# criar coluna bool empregado true/false

# 2.1 definir quais valores contam como false
# 2.2 atribuir true às demais linhas

df_p1['situacao'].value_counts()
```

```python
# 2.1
vfalse = ['Desempregado e não estou buscando recolocação',
          'Somente Estudante (pós-graduação)',
          'Prefiro não informar',
          'Somente Estudante (graduação)',
          'Desempregado, buscando recolocação'
          ]

print(11 + 25 + 16 + 68 + 205)
```

```python
# 2.2
df_p1['empregadx'] = df_p1['situacao'].apply(lambda x: False if x in vfalse else True)
df_p1['empregadx'].value_counts()
```

```python
# ajuste 3
# atribuir valores inteiros pra coluna salário

# 3.1 verificar .unique()
# 3.2 fazer atribuições
# 3.3 tratar os nan

# 3.1
df_p1['salario'].unique()
```

```python
# 3.2
df_p1['salario'] = df_p1['salario'].replace({
    'de R$ 1.001/mês a R$ 2.000/mês': 1500,
    'Menos de R$ 1.000/mês': 500,
    'de R$ 6.001/mês a R$ 8.000/mês': 7000,
    'de R$ 3.001/mês a R$ 4.000/mês': 3500,
    'de R$ 2.001/mês a R$ 3.000/mês': 2500,
    'de R$ 4.001/mês a R$ 6.000/mês': 5000,
    'de R$ 8.001/mês a R$ 12.000/mês': 10000,
    'de R$ 12.001/mês a R$ 16.000/mês': 14000,
    'de R$ 30.001/mês a R$ 40.000/mês': 35000,
    'de R$ 20.001/mês a R$ 25.000/mês': 22500,
    'de R$ 16.001/mês a R$ 20.000/mês': 18000,
    'de R$ 25.001/mês a R$ 30.000/mês': 27500,
    'Acima de R$ 40.001/mês': 40000
})

```

```python
# 3.3
df_p1['salario'] = df_p1['salario'].fillna(0).astype(int)
df_p1['salario'].info()
```
